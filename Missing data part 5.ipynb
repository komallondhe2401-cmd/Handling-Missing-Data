 # KNN Imputer

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('train.csv')
df.head()
```

```python
df=df[['Pclass','Age','Fare','Survived']]
df.head()
```

```python
df.isnull().sum()
```

```python
X=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
knn=KNNImputer(n_neighbors=3,weights='distance')
X_train_trf=knn.fit_transform(X_train)
X_test_trf=knn.transform(X_test)
```

```python
# Convert into dataframe
pd.DataFrame(X_train_trf,columns=X_train.columns)
```

```python
lr=LogisticRegression()
lr.fit(X_train_trf,y_train)
y_pred=lr.predict(X_test_trf)
accuracy_score(y_test,y_pred)
```

```python
# Comparision with Simple Imputer
si=SimpleImputer()

X_train_trf2=si.fit_transform(X_train)
X_test_trf2=si.transform(X_test)
```

```python
lr.fit(X_train_trf2,y_train)
y_pred=lr.predict(X_test_trf2)
accuracy_score(y_test,y_pred)
```

# Iterative Imputer

```python
df=pd.read_csv("50_Startups.csv")
df.head()
```

```python
df.columns
```

```python
df=df.drop(columns=['State'])
df=(df/10000).round()
np.random.seed(9)
print(df.sample(5))
```

```python

df=df.iloc[:,0:-1]
df.head()
```

```python
df.iloc[1,0]=np.nan
df.iloc[3,1]=np.nan
df.iloc[-1,-1]=np.nan
```

```python
df.head()
```

```python
# Step1 Replace all the Nan values with mean of that column
df0=pd.DataFrame()

df0['R&D Spend']=df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration']=df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend']=df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
df0=df0.round(2)
df0.head()
```

```python
# Removing the col1 imputed value
df1=df0.copy()
df1.iloc[1,0]=np.nan
df1.head()
```

```python
print(type(df1))
```

```python
df1.shape
```

```python
y=df1.iloc[[0,2,3,4],1]
y
```

```python
# The model work on non nan values to predict the nan value
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

# MICE using sklearn

```python
from sklearn.experimental import enable_iterative_imputer

from sklearn.impute import IterativeImputer
```

```python
data=np.array([
    [1,2,np.nan],
    [np.nan,4,7],
    [8,np.nan,10]
])
```

```python
imputer=IterativeImputer(max_iter=10,random_state=0)
X=imputer.fit_transform(data)
print(X)
```
